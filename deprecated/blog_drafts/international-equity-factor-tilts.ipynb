### Index Funds: The Evidence-Based Path to Wealth

Investing can be mind-numbingly complex, but one simple strategy has stood the test of time (for average retail investors such as myself): broad market index funds. Decades ago, economist Burton Malkiel boldly asserted (Malkiel, 2019) that ordinary investors are better off buying and holding a low-cost index fund than trying to pick stocks or hire expensive active managers. He backed it up with hard data: $10,000 invested in an S&P 500 index fund in 1969 grew to over $736,000 by 2014, dramatically outpacing the ~$501,000 result from the average actively managed fund (whose fees and trading shenanigans ate into returns). The takeaway? In the long run, the boring index fund investor came out about 50% richer than the fanciest fund-pickers on average. No stock-tipping wizardry required. Personal finance author JL Collins is on the same page. In *The Simple Path to Wealth* (Collins, 2016), Collins boils investing down to a few core rules: spend less than you earn, avoid debt, and invest the rest in low-cost index funds. It doesn't get much simpler (or more behaviourally sensible) than that. Collins basically says: you can become a millionaire just by consistently plowing money into a total market index fund (like Vanguard's VTI or VTSAX) and letting capitalism do its thing. Both Malkiel and Collins cite mountains of evidence that passive index investing beats most active strategies after costs. In other words, you're better off buying "the haystack" (the entire market) than searching for that one needle of a hot stock. Why do index funds win so often? Enter some foundational finance theory (as I understand it).


### The Science Behind Indexing (CAPM, EMH and Friends)

If the idea of efficient markets and modern portfolio theory makes you glaze over, don't worry, we'll keep this gentle (no pop quizzes on Greek letters for now, I promise). The key insight from Nobel-winning research of the 1950s-1960s is that diversification is one of the only free lunches in investing. Harry Markowitz showed that by holding a broad portfolio of assets, you can maximize your expected return for any given level of risk (Markowitz, 1952). He introduced a broad portfolio of assets, you can maximize your expected return for any given level of risk. He introduced the concept of an efficient frontier, which is basically the set of optimal portfolios that offer the highest return for a given risk level. The practical takeaway: don't put all your eggs in one basket, hold the whole market basket. By spreading out, you reduce risk without sacrificing return. Building on that, the Capital Asset Pricing Model (CAPM) took Markowitz's work a step further (Sharpe, 1964). Markowitz's work a step further. CAPM said if everyone diversifies perfectly, the market portfolio (i.e. an index of everything) is the optimal mix. In CAPM-world, a stock's expected return relates to how much it moves with the market (its beta). The market itself is the only risk factor you get paid for; everything else (company-specific ups and downs) gets diversified away. In plainer terms, CAPM implies that the market index is king – it's the most efficient, and the only thing that reliably drives returns. Trying to beat the market by picking stocks or timing sectors is, on average, a zero-sum game. For every winner, there's a loser, especially after fees.This ties nicely to the Efficient Market Hypothesis (EMH), articulated by Eugene Fama in 1970 (Fama, 1970). EMH argues that all available information is quickly baked into asset prices. If that's true, consistently outsmarting the market is about as easy as consistently flipping heads on a coin. Prices move in a "random walk", essentially unpredictable, because any new info (earnings surprise, Fed decision, you name it) gets rapidly absorbed by thousands of rational (and not-so-rational) market participants.

The implication for us average retail investors: you can't reliably find underpriced gems, so don't waste time trying. Instead, just buy an index fund and get the market return, which by definition beats most investors after costs. EMH isn't perfect (markets can get a little crazy at times), but by and large even professional fund managers struggle to beat a simple index over the long haul. So far we've covered that index funds, capturing the broad market, are backed by strong theory (CAPM, EMH) and empirical results. You get market beta (the return of the market) cheap and easy, which historically has been quite generous over decades. As Malkiel quips, a broad index fund guarantees you'll always get the average market return,  and since the market is the average of all investors, being average actually puts you ahead of most, once you factor in the hefty fees and mistakes others make. But (there's always seemingly a "but" in finance), academics kept digging. They found there might be more to returns than just the market. Not all "average" is created equal, which brings us to factor investing.


### Factor Investing 101 (A "Smart Beta" Extension of CAPM)

Let's say you've embraced the zen of index funds, be the market, my friend. Does that mean there's no way to tweak things for potentially better returns (at acceptably higher risk)? Enter factor investing, sometimes branded as "smart beta" (a marketing term Jack Bogle himself snarked was "smart beta is stupid" – he didn't mince words on this topic). In more polite terms, factor investing is an extension of the CAPM that recognizes multiple drivers of returns beyond just market beta. Let us think of the market like an all-you-can-eat buffet. CAPM said "just fill your plate with a bit of everything, that's optimal." Factor investing says "well, maybe some sections of the buffet consistently give more nutrition per calorie." In fact, Columbia professor Andrew Ang famously put it this way: "Factors are to assets what nutrients are to food" (Ang, 2014). Stocks and portfolios are just bundles of underlying "nutrients" (risk factors) that drive returns. Instead of just accepting the whole market basket as-is, factor investors tilt their portfolios toward certain factors, traits like size, value, momentum, quality/profitability, etc (where understanding factors and their possible decay over time would and probably will be a whole other post of mine), that research has shown offer long-term return premiums. Where did these ideas come from? In 1992, Fama and French found that two simple characteristics, a stock's size (small vs. large) and valuation (cheap "value" stocks vs. expensive "growth" stocks), explained a lot of differences in stock returns that the market alone didn't (Fama & French, 1992). simple characteristics, a stock's size (small vs. large) and valuation (cheap "value" stocks vs. expensive "growth" stocks), explained a lot of differences in stock returns that the market alone didn't. Historically, smaller companies and value stocks delivered higher returns than the market, albeit with their own roller-coaster rides. Later research added more "nutrients" to the salad: momentum (stocks that have been going up tend to keep going up briefly), profitability/quality (more profitable firms' stocks do better), low volatility, etc. By now there's a whole zoo of factors identified in academia, but as author Larry Swedroe notes, you only need a handful of the major ones to explain most of the differences in portfolio performance (Swedroe & Berkin, 2016). In fact need a handful of the major ones to explain most of the differences in portfolio performance. In fact, Swedroe and co-author Andrew Berkin suggest focusing on just five equity factors: the market itself, size, value, profitability/quality, and momentum. (Yes, the factor zoo has a "big 5" – like a basketball lineup or the Spice Girls, depending on your preferred metaphor). In plainer language: factor investing means tilting your portfolio toward certain types of stocks that, over the long run, have rewarded investors with higher returns (for bearing some extra risk or exploiting behavioral tendencies). It's grounded in tons of data (and theory). For example, value stocks (those with low prices relative to fundamentals) have on average beaten growth stocks over many decades in many markets, albeit with painfully long dry spells. Small-cap stocks (think scrappy little companies) have outpaced large-caps over the long run (again, not every year). These premiums are believed to be compensation for risk (e.g. small and value companies are often riskier, out of favor, or harder to arbitrage) and/or investor behavioral biases. 

Even Burton Malkiel, in the 50th Anniversary edition of *A Random Walk*, acknowledges that funds targeting these factors – like Dimensional Fund Advisors (DFA) funds – have delivered excess returns in the past. He points out, for instance, that DFA's small-cap value fund beat the market by over 1% annually in one 10-year stretch. However, and this is a professorly however (as professorly as a non-academic can get), he cautions that these higher returns are likely just compensation for higher risk, and "smart beta" strategies can go through long periods of underperformance. Indeed, all factors have rough decades (hello 2010s, when value looked dead). So Malkiel's advice is usually to keep the core of your portfolio in broad index funds, and only tilt to factors cautiously, if at all. (He basically says, if you must gamble on a factor, do it prudently around the edges, and don't pay high fees for it). To sum up: Index funds = owning the haystack (market beta) at minimal cost, backed by CAPM and EMH. Factor investing = adding a bit of spice by overweighting certain "nutrients" in the haystack (size, value, etc.) that may deliver extra return. It's an evolution, not a revolution, built on the same idea of disciplined, diversified investing – just with a rules-based active twist. People like Swedroe, Ang, and DFA's founders (David Booth, Eugene Fama & Ken French, among others) have championed this approach as a way to potentially improve on plain indexing, without venturing into the dark arts of stock picking or market timing. Now, all this theory is nice, but how does an investor actually implement it, or parts of it? That brings us to my personal journey (and mild obsession): moving from the classic Bogleheads three-fund portfolio toward a portfolio that is an intermediary between pure market beta and factor tilting, step by step, and hitting an annoying snag in the process, finding a worthy replacement for VXUS (Vanguard's total international index). 


### The Three-Fund Portfolio (Bogleheads Basics)

First stop: the beloved Three-Fund Portfolio. This is the baby Yoda of personal finance – small, simple, and wise. The three-fund portfolio typically consists of: 

* (1) a total U.S. stock market index fund (like Vanguard Total Stock Market, ticker VTI; Vanguard, 2024), 
* (2) a total international stock index fund (like Vanguard Total International, ticker VXUS; Vanguard, 2024)
* (3) a total U.S. bond market fund (like BND).
  
 That's it. With just three funds, you get exposure to literally thousands of stocks around the globe and the broad U.S. investment-grade bond universe. It's hard to overstate how much diversification and simplicity you get in this minimalist setup. Legendary Bogleheads like Taylor Larimore (Bogleheads Wiki, 2024) (co-author of *The Bogleheads' Guide*) tout the three-fund portfolio's many virtues: extremely low cost, broad diversification, no overlapping holdings, minimal maintenance, and no reliance on any star fund manager. You effectively own the entire world's public markets in one neat package, at a fee that's often 0.05% or less per year. Tough to beat. Of course, you still have to decide your allocation, i.e. what percentage in each of the three funds. This usually comes down to your risk tolerance and goals. Stocks vs bonds is the big lever (more bonds = safer, but lower long-term returns). And within stocks, how much U.S. vs international? Reasonable people (and unreasonable ones on internet forums) debate the "perfect" ratio. Vanguard's research suggests anywhere from 20% to 50% of stocks in international is reasonable, with ~40% being market-weight. Jack Bogle himself and the earlier cited JL Collins were both famously lukewarm on international, preferring an 80/20 or even 100/0 home bias (he believed U.S. companies already have global exposure). Meanwhile, others argue for market-cap weights (~60% US, 40% international currently). In practice, many Bogleheads split the difference. For example, one common recommendation for a balanced portfolio is 50% VTI (U.S.), 20% VXUS (international), 30% BND (bonds). That would be a 70/30 stock/bond mix, with international making up ~29% of stocks. Another example (from Vanguard's own target-risk funds for retirees) is ~40% VTI, 20% VXUS, 40% BND (so 60/40 with 1/3 of stocks international). There's no single right answer, but the key is these portfolios are broadly diversified and simple. For our purposes here, I'm focusing on the equity portion (the stocks), since that's where the factor intrigue comes in. Bonds (BND) are important, but they're not part of the "factor investing" vs "index" debate for now (in this post specifically); so we will simplify and say they're just for safety/interest. So let's set BND aside and talk stocks: VTI + VXUS as the classic equity allocation. How has a basic VTI/VXUS global stock combo performed? Quite well historically, albeit with some U.S. vs international performance cycles that tempt investors to stray. To visualize this, let's do a quick backtest with some code. We'll compare different mixes of VTI and VXUS, say from all-U.S. to all-international, and see the historical performance. (Don't worry, we'll exclude bonds for now, keeping it apples-to-apples.)


```python
# VTI/VXUS Growth Comparison (Net of Expense Ratios)
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Load pre-downloaded data
prices = pd.read_csv('../data/prices_data.csv', index_col=0, parse_dates=True)
prices = prices[['VTI', 'VXUS']].loc['2011-01-28':'2024-12-31']

# Daily returns
returns = prices.pct_change().dropna()

# Expense ratios (annual) - deducted daily
EXPENSE_RATIOS = {'VTI': 0.0003, 'VXUS': 0.0007}  # 0.03% and 0.07%
daily_cost = {k: v / 252 for k, v in EXPENSE_RATIOS.items()}

# Net returns after fees
returns_net = returns.copy()
for ticker in ['VTI', 'VXUS']:
    returns_net[ticker] = returns[ticker] - daily_cost[ticker]

# Portfolio allocations
allocations = {
    '100% US (VTI)': {'VTI': 1.0, 'VXUS': 0.0},
    '60/40 US/Intl': {'VTI': 0.6, 'VXUS': 0.4},
    '100% Intl (VXUS)': {'VTI': 0.0, 'VXUS': 1.0}
}

# Calculate cumulative growth
growth_df = pd.DataFrame({
    name: (1 + returns_net['VTI'] * w['VTI'] + returns_net['VXUS'] * w['VXUS']).cumprod()
    for name, w in allocations.items()
})

# Print summary
print('Growth of $1 (Net of Expense Ratios, 2011-2024)')
print('=' * 50)
for col in growth_df.columns:
    final = growth_df[col].iloc[-1]
    print(f'{col:<20}: ${final:.2f} ({(final-1)*100:+.1f}%)')

# Static plot - clean, minimal, blog-ready
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(10, 5.5), dpi=120)

colors = {'100% US (VTI)': '#2563eb', '60/40 US/Intl': '#7c3aed', '100% Intl (VXUS)': '#059669'}

for col in growth_df.columns:
    style = {'linewidth': 2.2, 'linestyle': '--'} if '60/40' in col else {'linewidth': 1.8, 'linestyle': '-'}
    ax.plot(growth_df.index, growth_df[col], label=col, color=colors[col], **style)
    ax.annotate(f'${growth_df[col].iloc[-1]:.2f}', xy=(growth_df.index[-1], growth_df[col].iloc[-1]),
                xytext=(8, 0), textcoords='offset points', fontsize=9, color=colors[col], fontweight='medium')

ax.set_title('Growth of $1: US vs International Equity Mix\n(2011-2024, Net of Expense Ratios)', 
             fontsize=13, fontweight='bold', pad=12)
ax.set_ylabel('Portfolio Value ($)', fontsize=11)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${x:.0f}'))
ax.xaxis.set_major_locator(mdates.YearLocator(2))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax.legend(loc='upper left', frameon=True, framealpha=0.95, fontsize=10, edgecolor='none')
ax.spines[['top', 'right']].set_visible(False)
ax.grid(True, alpha=0.3, linewidth=0.5)
plt.tight_layout()
plt.show()
```

(The above code fetches monthly prices for VTI and VXUS, then plots how $1 would have grown in different blends. We highlight the 60/40 (US/Intl) mix as a commonly suggested allocation.)


Running this, you'd clearly see that 100% US (VTI) has led the pack in the last decade (U.S. stocks dominated the 2010s), while 100% international (VXUS) lagged. The 60/40 mix falls in between. If this were 2000-2009, the plot would tell the opposite story (that was a lost decade for U.S. stocks and a decent one for international). The lesson: global diversification smooths out regional swings, when U.S. stocks zig, internationals often zag. The 60% US / 40% international mix (roughly global market cap weight) is a nice balanced approach, and indeed we see it avoids the extremes – it's never the very top or bottom performer in any period. That balanced portfolio will be our baseline going forward. Now, as a die-hard Boglehead, you could stop right here. "Two stock funds (plus a bond fund) and chill, as I often see cited in the bogleheads subreddit" is a perfectly fine mantra! So why even consider changing it? Well, recall the earlier discussion: index funds have minor inefficiencies and there might be systematic ways to squeeze out a bit more return. Which brings me to Dimensional Fund Advisors and their shiny "Market-Series" fund offerings like [DFUS](https://www.dimensional.com/us-en/our-approach/dimensional-equity-solutions).


### VTI vs DFUS (Index Funds, Upgraded?)

Imagine you love your Honda Civic (reliable, efficient, gets you there) but you hear about a new model that's just as cheap on gas yet somehow a tad faster and smoother because the engineers tuned it differently. That's essentially DFUS vs VTI. VTI is the classic total U.S. stock market ETF – it just buys everything per the index. DFUS (Dimensional U.S. Equity Market ETF) is a relatively new entrant (launched 2021) that covers the same broad U.S. market but doesn't strictly follow an index. It's actively managed in a very rules-based way. Think of DFUS as VTI's nerdy cousin who read all the finance journals and decided to avoid a few quirks of traditional index funds.What quirks, you ask? The folks at DFA identified a few things: (1) Traditional index funds have to periodically rebalance when index membership changes, which can be front-run by traders. (For example, when the S&P 500 announces a new stock addition, index funds must buy it en masse at a potentially inflated price – DFA's active approach can be more patient or avoid overpriced newcomers.) (2) Market indexes often include some low-profitability small growth stocks (historically poor performers) and certain high-turnover sectors like REITs that can be tax-inefficient. DFUS's strategy avoids or underweights some of these segments. In fact, DFUS was converted from a legacy tax-managed fund, so it deliberately excludes REITs (Real Estate Investment Trusts) and some tiny unprofitable growth companies for tax and expected-return reasons. REITs were only ~3% of VTI, but notably they've lagged lately, so not holding them gave DFUS a boost. Similarly, by not holding the worst of small/growth "junk" stocks, DFUS skews slightly toward higher-quality or value names, which can improve returns (and incidentally tilts the portfolio a hair more toward large growth, since it's skipping the small growth).Crucially, DFUS still holds all sizes of stocks, weighted roughly by market cap, to maintain broad diversification. It's not making big concentrated bets – it owns ~2,400 stocks, from Apple down to micro-caps. It's just not forced to mirror an index committee's list. This flexibility lets DFA's algorithms be opportunistic around trading: e.g., they can delay buying a stock that's shooting up (avoiding "buy high" scenarios forced by index inclusion timing), or they can harvest tax losses more aggressively. DFUS also tends to be very tax-efficient (no capital gain distributions so far), partly thanks to the ETF structure plus active loss-harvesting and in-kind redemptions that DFA is good at.The result? Since its mid-2021 inception as an ETF, DFUS has actually outperformed VTI (same period). It's only a couple years of data, so don't uncork the champagne just yet, but it's encouraging. Ben Felix (of Rational Reminder fame) did an analysis on this: he compared DFUS to VTI to illustrate how "not following an index can be beneficial". Despite both funds being ultra-diversified U.S. market vehicles, DFUS beat VTI by a modest margin (in large part due to those exclusions of REITs and some structural tweaks). Importantly, DFUS achieves this without charging an arm and a leg – its expense ratio is ~0.09%, only a few basis points higher than VTI's ~0.03%. For that tiny fee difference, you get DFA's secret sauce of systematic trading and factor-informed tweaks.


As an investor who wants to "do slightly better if possible, but never do worse," DFUS is intriguing (DFA, 2024; Felix, 2024). It's basically VTI 2.0 – same broad exposure, just optimized a bit on the margins. Even DFA's own description is straightforward (DFA, 2024): "The fund actively selects US equities of all sizes weighted by market cap, seeking long-term appreciation." In other words, it's not deviating wildly from the market – just trying to be smart about the small stuff.Let's see how DFUS vs VTI have actually performed with some numbers. We'll fetch historical data and compare returns:


```python
# VTI vs DFUS Comparison (Net of Expense Ratios)
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Load data
df = pd.read_csv('../data/prices_data.csv', index_col=0, parse_dates=True)
start_date = df['DFUS'].first_valid_index()

# Daily returns
vti_ret = df.loc[start_date:, 'VTI'].pct_change().dropna()
dfus_ret = df.loc[start_date:, 'DFUS'].pct_change().dropna()

# Expense ratios (annual) - deducted daily
EXPENSE_RATIOS = {'VTI': 0.0003, 'DFUS': 0.0009}  # 0.03% and 0.09%
daily_cost = {k: v / 252 for k, v in EXPENSE_RATIOS.items()}

# Net returns & cumulative growth
vti_growth = pd.concat([pd.Series([1.0], index=[start_date]), (1 + vti_ret - daily_cost['VTI']).cumprod()])
dfus_growth = pd.concat([pd.Series([1.0], index=[start_date]), (1 + dfus_ret - daily_cost['DFUS']).cumprod()])

# Print summary
print(f'Performance since DFUS inception ({start_date.date()}):')
print('=' * 50)
print(f"VTI Total Return:  {(vti_growth.iloc[-1]-1)*100:+.1f}%")
print(f"DFUS Total Return: {(dfus_growth.iloc[-1]-1)*100:+.1f}%")
print(f"DFUS Outperformance: {(dfus_growth.iloc[-1]-vti_growth.iloc[-1])*100:+.1f}%")

# Static plot
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(10, 5.5), dpi=120)

colors = {'VTI': '#2563eb', 'DFUS': '#dc2626'}
ax.plot(vti_growth.index, vti_growth, label='VTI (Total US Market)', color=colors['VTI'], linewidth=1.8)
ax.plot(dfus_growth.index, dfus_growth, label='DFUS (DFA US Equity)', color=colors['DFUS'], linewidth=2.2, linestyle='--')

for growth, color in [(vti_growth, colors['VTI']), (dfus_growth, colors['DFUS'])]:
    ax.annotate(f'${growth.iloc[-1]:.2f} ({(growth.iloc[-1]-1)*100:+.1f}%)', 
                xy=(growth.index[-1], growth.iloc[-1]), xytext=(8, 0), 
                textcoords='offset points', fontsize=9, color=color, fontweight='medium')

ax.set_title(f'VTI vs DFUS: Growth of $1 Since DFUS Inception\n({start_date.strftime("%b %Y")}–Present, Net of Expense Ratios)', 
             fontsize=13, fontweight='bold', pad=12)
ax.set_ylabel('Portfolio Value ($)', fontsize=11)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${x:.2f}'))
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.xticks(rotation=45, ha='right')
ax.legend(loc='upper left', frameon=True, framealpha=0.95, fontsize=10, edgecolor='none')
ax.spines[['top', 'right']].set_visible(False)
ax.grid(True, alpha=0.3, linewidth=0.5)
plt.tight_layout()
plt.show()
```

(The above will give total growth of $1 in each fund since DFUS's launch. I expect DFUS to be a bit higher. We could also chart the difference over time.)

The data likely shows DFUS pulling ahead of VTI by a few percentage points cumulatively. For instance, if VTI returned say +30% over the period, DFUS might be +33% (just illustrative). Not life-changing, but hey, every bit counts, especially if achieved efficiently. The key is DFUS did it while providing the same broad market exposure and volatility. It didn't require genius stock picking or clairvoyant timing – just some rules-based adjustments (like leaving REITs off the guest list and doing more tax-efficient trading).At this point, I was pretty sold on replacing VTI with DFUS in my own portfolio for U.S. equity exposure. It's a classic free (or cheap) lunch story: essentially the same exposure, potential for slightly better returns, slightly higher tax efficiency, for nearly the same low fee. As one wit on the Bogleheads forum put it, "DFUS fixes some academically researched inefficiencies" with index funds. And indeed, Dimensional's whole ethos (since being founded in 1981) is "apply academic research to real-world portfolios." They were doing factor investing before it even had a cool name – Fama and French have long ties to the firm. For decades DFA offered only mutual funds, often accessible only through advisors, focusing on small-cap and value tilts. In 2020-21 they finally jumped into ETFs, converting a bunch of funds. DFUS is one such conversion (hence the long live track record back to 2001 in its stats). The history and approach of DFA is a fun story: started by alumni of the University of Chicago's finance department, they partnered with academics to engineer portfolios capturing size, value, etc., and insisted on evidence-based investing. They also pioneered things like patient trading (not chasing prices), block buying from liquidity-seeking sellers, and avoiding the index front-running problem. All that carries into DFUS.Alright, U.S. stocks solved – just swap VTI for DFUS, done! Now onto the international side. This is where my plan ran into a snag…


### The International Challenge

Vanguard's VXUS (Total International Stock ETF) is the international analog to VTI: a cap-weighted index of all non-U.S. stocks in both developed and emerging markets. VXUS is cheap (~0.07% expense) and very broad (around 7,500 stocks in ~45 countries). If I want to replicate my DFUS logic abroad, I'd look for a DFA ETF that does "total international" with similar slight improvements. However, no single DFA ETF is a perfect VXUS equivalent at the moment, especially not with the ultra-low fee and simplicity of DFUS.Here's the landscape: DFA chose to split the world ex-US into two pieces – one for developed markets and one for emerging markets. These are DFAI (Dimensional International Core Equity Market ETF) and DFAE (Dimensional Emerging Core Equity Market ETF). Both are actively managed, factor-oriented funds (the "Core Equity" name is a hint that they tilt toward size/value/profitability, akin to DFA's classic strategies). To mirror VXUS (which is ~75% developed, ~25% emerging by market cap), you'd have to hold both DFAI and DFAE in roughly that ratio. That's doable, but it's more complicated than a single fund, and crucially the fees are higher. As of now, DFAI's expense ratio is around ~0.18% and DFAE's is ~0.36% (emerging markets funds tend to cost more). If you blend 75/25, the weighted expense comes out about ~0.225% – let's call it 0.23%. That's triple VXUS's fee. Not outrageous, but it gives one pause when VXUS is so darn cheap.What about performance and approach? DFAI and DFAE are tilted funds. For example, according to DFA (and Morningstar), DFAI invests in companies of all sizes in developed markets, but with "increased exposure to smaller capitalization, lower relative price, and higher profitability" stocks. Likewise, DFAE targets emerging market stocks with an emphasis on value and profitability. They are not trying to match the index; they're trying to beat it by systematically overweighting those factors. Over long periods, I actually expect they could outperform VXUS (just as DFA's other factor funds have often beaten cap-weight indices historically). But they will also deviate more from VXUS in the short term because of these tilts. It's a bit more "spicy" than DFUS, which was pretty close to the market. DFAI+DFAE is more like adding hot sauce – tasty potential, but it might upset your stomach once in a while.To be fair, Dimensional does offer one single ETF for international: DFAX (Dimensional World ex-US Core 2 ETF). DFAX does bundle developed and emerging in one fund (so it's closer to a VXUS counterpart). But two issues: (1) Its expense ratio is ~0.29%, even higher (because "Core 2" means even heavier tilts towards small/value – more trading, more costs). And (2) it's actively tilting quite a bit, so it's definitely not a plain-vanilla index approach. DFAX is essentially the next step on the factor continuum – a full-throated factor fund for international stocks. If DFUS is "market with a twist of lime," DFAX is more like "handcrafted cocktail with 3 shots of factor tequila."


For a moment, though, let's see how DFAX has done vs VXUS. Perhaps the extra oomph has paid off so far. I'll pull some data and even attempt a mini factor regression to illustrate differences:


```python
# VXUS vs DFAX Comparison (Net of Expense Ratios)
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Load data
df = pd.read_csv('../data/prices_data.csv', index_col=0, parse_dates=True)
start_date = df['DFAX'].first_valid_index()

# Daily returns
vxus_ret = df.loc[start_date:, 'VXUS'].pct_change().dropna()
dfax_ret = df.loc[start_date:, 'DFAX'].pct_change().dropna()

# Expense ratios (annual) - deducted daily
EXPENSE_RATIOS = {'VXUS': 0.0007, 'DFAX': 0.0023}  # 0.07% and 0.23%
daily_cost = {k: v / 252 for k, v in EXPENSE_RATIOS.items()}

# Net returns & cumulative growth
vxus_growth = pd.concat([pd.Series([1.0], index=[start_date]), (1 + vxus_ret - daily_cost['VXUS']).cumprod()])
dfax_growth = pd.concat([pd.Series([1.0], index=[start_date]), (1 + dfax_ret - daily_cost['DFAX']).cumprod()])

# Print summary
print(f'Performance since DFAX inception ({start_date.date()}):')
print('=' * 50)
print(f"VXUS Total Return: {(vxus_growth.iloc[-1]-1)*100:+.1f}%")
print(f"DFAX Total Return: {(dfax_growth.iloc[-1]-1)*100:+.1f}%")
print(f"DFAX Outperformance: {(dfax_growth.iloc[-1]-vxus_growth.iloc[-1])*100:+.1f}%")

# Static plot
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(10, 5.5), dpi=120)

colors = {'VXUS': '#059669', 'DFAX': '#7c3aed'}
ax.plot(vxus_growth.index, vxus_growth, label='VXUS (Total Intl Market)', color=colors['VXUS'], linewidth=1.8)
ax.plot(dfax_growth.index, dfax_growth, label='DFAX (DFA World ex-US)', color=colors['DFAX'], linewidth=2.2, linestyle='--')

for growth, color in [(vxus_growth, colors['VXUS']), (dfax_growth, colors['DFAX'])]:
    ax.annotate(f'${growth.iloc[-1]:.2f} ({(growth.iloc[-1]-1)*100:+.1f}%)', 
                xy=(growth.index[-1], growth.iloc[-1]), xytext=(8, 0), 
                textcoords='offset points', fontsize=9, color=color, fontweight='medium')

ax.set_title(f'VXUS vs DFAX: Growth of $1 Since DFAX Inception\n({start_date.strftime("%b %Y")}–Present, Net of Expense Ratios)', 
             fontsize=13, fontweight='bold', pad=12)
ax.set_ylabel('Portfolio Value ($)', fontsize=11)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${x:.2f}'))
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.xticks(rotation=45, ha='right')
ax.legend(loc='upper left', frameon=True, framealpha=0.95, fontsize=10, edgecolor='none')
ax.spines[['top', 'right']].set_visible(False)
ax.grid(True, alpha=0.3, linewidth=0.5)
plt.tight_layout()
plt.show()

```

(Above, we'd see if DFAX outperformed VXUS since it launched. I suspect yes, given DFA's tilts have helped in recent value-friendly years. Suppose VXUS did +15% and DFAX +20% cumulatively – just making up numbers.)

Now, about that factor regression: In an ideal world, I'd run a multi-factor regression on both VXUS and DFAX returns to quantify their exposures (betas) to, say, the MSCI World-exUS Index (market), a size factor, a value factor, etc. Without diving into data libraries, I can tell you the likely outcome: VXUS = basically pure market beta (by design, it's cap-weighted). DFAX = market beta + a hefty tilt to value and small. In plain English, DFAX takes more of its risk in smaller and cheaper stocks. That's been good recently (2022-2023 were decent for value globally), hence DFAX's outperformance. But it could lag if large growth stocks roar back or if emerging markets (which DFAX might overweight) stumble relative to developed.From a practical perspective, if I were ready to fully embrace factor investing for equities, I might pair DFAC (Dimensional US Core 2, a U.S. factor ETF) with DFAX (World ex-US Core 2) for a complete globally tilted portfolio (those two tickers even sound like they belong together). In fact, DFAC+DFAX is essentially DFA's "equity core" solution – high expected returns, higher factor exposure, still broadly diversified. However – and this is a Jim Carrey "big gulps huh? well, see ya later" kind of however – I'm not quite ready to go "full factor" with all my equity money. Tilting adds tracking error (short-term differences from the market) that can test one's patience and fortitude. As Malkiel would remind us, not every factor will deliver on schedule. If I tilt too much and hit a decade of underperformance, will I lose faith and capitulate at the worst time? Possibly. That's an important self-check.Thus, my current plan is incremental improvement: start with the easy win (DFUS over VTI for U.S.), and wishfully look for an easy win on the international side. Unfortunately, that "VXUS 2.0" hasn't materialized in the form of a single DFA or Avantis ETF with broad market coverage, mild tilts, and low fee.


Avantis Investors, another shop I admire (founded by ex-DFA folks; Avantis, 2024, by the way), also offers international factor funds – but similarly split: AVDE (developed int'l, 0.23% fee; Avantis, 2024) and AVEM (emerging, 0.33% fee; Avantis, 2024). They too expect you to mix your own VXUS cocktail. No simple one-ticker solution there either. So either route, DFA or Avantis, I'm stuck with a two-fund combo for international, with higher fees and more tracking error.Why doesn't a DFA "DFIX" (Dimensional International Equity Market ETF) exist, analogous to DFUS? My guess: DFA's philosophy is that outside the U.S., the benefits of tilting are even more compelling (and perhaps the costs of indexing even higher), so they didn't bother with a plain "market" version. International markets are less efficient in some cases, and DFA probably figures investors going with them want the tilts. Another reason might be that their mutual fund conversions naturally split that way (they had separate funds for International and Emerging originally, no single total international mutual fund to convert). Meanwhile, Vanguard's VXUS is so cheap that competing directly on a plain cap-weighted ex-US index with similar fee might not be profitable for DFA/Avantis. Instead, they differentiate by offering the factor angle.As a result, I find myself in a conundrum: I can't get a DFUS-like upgrade for VXUS without accepting higher fees or more complexity. This bothers the optimizer in me! It's like having a brand new iPhone in the U.S. portion of my portfolio but using a reliable, old Nokia brick for the international part – it works, but I know there's something sleeker out there.For now, my plan is to stick with VXUS for international while I continue researching and gaining comfort with the DFAI/DFAE or DFAX approach. I might dip a toe by allocating a portion of international to DFA's funds to see how they behave. If factor tilts keep proving their worth (and I don't lose sleep over the differences), I could eventually transition fully: US equities in DFAC (a bit spicier than DFUS) and international in DFAX. That would effectively turn the plain three-fund portfolio into a three-fund factor portfolio (DFAC, DFAX, and some bond fund). But I'm not there yet emotionally or intellectually (still reading white papers into the wee hours, as one does).


## Summary and Conclusion

In summary, here's the journey so far:
- **Start**: Bogleheads 3-fund (VTI, VXUS, BND). Hard to criticize – it's evidence-based and incredibly simple.
  
- **Step 1**: Upgrade U.S. stocks by swapping VTI -> DFUS, to capture a bit more return via DFA's magic, without altering the core exposure (market beta still intact). This change is stealthy – most casual observers wouldn't notice a difference, except perhaps a slightly juicier 1099-DIV or a mildly higher end balance in a few years.
  
- **Step 2**: Attempt to upgrade international stocks. Realization: no straightforward DFUS analog exists. Options are DFAI+DFAE or DFAX (or Avantis AVDE+AVEM). All involve higher cost and more tracking error. The benefits are there (historically DFA's international value and small tilts have added value), but the leap is bigger.

- **Step 3 (future?)**: Once I fully drink the factor Kool-Aid (factor-aid?), go "all-in" with DFAC + DFAX to replace VTI + VXUS, completing the transformation to a factor-tilted global portfolio. That would give me greater small/value exposure across the board, hopefully boosting long-run returns and diversification. But it also means I must endure periods when I lag a vanilla global index – and that's the price of admission for factor premiums.I'll admit, part of me feels like the dog who caught the car – I wanted to optimize every part of my portfolio, and now I'm annoyed that I can't easily optimize VXUS the same way I did VTI. Perhaps patience is warranted; DFA might someday launch an "International Market ETF" with lower fee and mild tilts if demand is there. Or Avantis could bundle developed+emerging into one fund (they haven't yet, maybe due to similar reasons as DFA).Until then, I'll practice what John Bogle preached: Don't let the better be the enemy of the good. VXUS is already very good. I'll keep holding it, while enjoying the incremental upgrade DFUS provides on the U.S. side. And I'll keep an eye on those DFAI/DFAE funds – who knows, maybe I'll gradually add some in and become the very international "optimizer" I swore I wouldn't be. If you see me on a street corner muttering about expense ratios and factor loadings, well, now you know my story.Investing, like life, is a journey – sometimes from simplicity to complexity and back again. My current stop is somewhere in between, armed with index funds in one hand and factor funds in the other, hoping I've struck a reasonable balance.In the end, whether you're a Boglehead purist or a factor enthusiast, remember that sticking with a sound strategy is more important than choosing the "perfect" strategy. As Malkiel and Collins would remind us: get the big things right (diversification, low costs, long-term discipline), and you're already way ahead. Any further optimizations are just icing on the cake (or extra nutrients in the soup, as Andrew Ang might say). And who knows – by the time you read this, maybe that elusive VXUS replacement will exist, and I'll be first in line to buy it. Until then, I'll keep Thinking Things Through (true to my blog's title) and enjoying the process of learning – hopefully without straying too far from the simple path that got me here.Cheers to continuous improvement, and may our portfolios be ever efficient (if not perfectly optimal)!
