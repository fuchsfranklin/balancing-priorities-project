````python
# Deep Dive Analysis: Extract Fascinating Portfolio Insights
print("🔍 Performing comprehensive analysis of all processed data...")

# Analyze portfolio composition patterns
print("\n📊 PORTFOLIO COMPOSITION ANALYSIS:")

# Top 5 best Sharpe ratio portfolios and their allocations
top_performers = pareto_results_df.nlargest(5, 'sharpe_ratio')
print(f"🏆 TOP 5 PERFORMING PORTFOLIOS (by Sharpe ratio):")

fascinating_insights = {
    'portfolio_compositions': {},
    'asset_patterns': {},
    'performance_gaps': {},
    'diversification_insights': {},
    'cost_efficiency': {}
}

for i, (idx, portfolio) in enumerate(top_performers.iterrows()):
    portfolio_weights = pareto_portfolios_df.loc[idx]
    non_zero_weights = portfolio_weights[portfolio_weights > 0.01]  # Show weights > 1%
    
    print(f"\n   Portfolio #{i+1} (Sharpe: {portfolio['sharpe_ratio']:.3f}):")
    print(f"   Return: {portfolio['return']:.1%} | Risk: {portfolio['risk']:.1%} | Cost: {portfolio['cost']:.3%}")
    
    composition_str = ""
    for asset, weight in non_zero_weights.sort_values(ascending=False).items():
        composition_str += f"   • {asset}: {weight:.1%}\n"
    print(composition_str)
    
    fascinating_insights['portfolio_compositions'][f'top_{i+1}'] = {
        'sharpe': portfolio['sharpe_ratio'],
        'composition': non_zero_weights.to_dict(),
        'return': portfolio['return'],
        'risk': portfolio['risk'],
        'cost': portfolio['cost']
    }

# Analyze asset allocation patterns across ALL portfolios
print(f"\n📈 ASSET ALLOCATION PATTERNS (across {len(pareto_portfolios_df)} portfolios):")
asset_stats = {}
for asset in pareto_portfolios_df.columns:
    weights = pareto_portfolios_df[asset]
    non_zero_weights = weights[weights > 0]
    
    asset_stats[asset] = {
        'avg_weight': weights.mean(),
        'max_weight': weights.max(),
        'usage_frequency': (weights > 0.01).sum() / len(weights),  # Frequency of meaningful allocation (>1%)
        'median_when_used': non_zero_weights.median() if len(non_zero_weights) > 0 else 0
    }
    
    if asset_stats[asset]['usage_frequency'] > 0.1:  # Show assets used in >10% of portfolios
        print(f"   {asset}: Used in {asset_stats[asset]['usage_frequency']:.1%} of portfolios")
        print(f"     → Avg allocation: {asset_stats[asset]['avg_weight']:.1%}")
        print(f"     → Max allocation: {asset_stats[asset]['max_weight']:.1%}")
        print(f"     → Typical allocation when used: {asset_stats[asset]['median_when_used']:.1%}")

fascinating_insights['asset_patterns'] = asset_stats

# Compare with traditional Bogleheads approaches
print(f"\n⚡ TRADITIONAL vs PARETO PERFORMANCE GAP:")
best_pareto_sharpe = pareto_results_df['sharpe_ratio'].max()
best_trad_sharpe = bogleheads_comparison_df['sharpe_ratio'].max()
worst_trad_sharpe = bogleheads_comparison_df['sharpe_ratio'].min()

performance_gap = (best_pareto_sharpe - best_trad_sharpe) / best_trad_sharpe * 100

print(f"   🏆 Best Pareto Sharpe: {best_pareto_sharpe:.3f}")
print(f"   📊 Best Traditional Sharpe: {best_trad_sharpe:.3f}")
print(f"   💥 Performance Improvement: {performance_gap:.1f}%")
print(f"   📉 Traditional Range: {worst_trad_sharpe:.3f} - {best_trad_sharpe:.3f}")
print(f"   📈 Pareto Range: {pareto_results_df['sharpe_ratio'].min():.3f} - {pareto_results_df['sharpe_ratio'].max():.3f}")

fascinating_insights['performance_gaps'] = {
    'best_pareto': best_pareto_sharpe,
    'best_traditional': best_trad_sharpe,
    'improvement_pct': performance_gap,
    'pareto_range': [pareto_results_df['sharpe_ratio'].min(), pareto_results_df['sharpe_ratio'].max()],
    'traditional_range': [worst_trad_sharpe, best_trad_sharpe]
}

# Load and analyze correlation data
asset_correlations_df = pd.read_csv(processed_data_dir / 'phase7_asset_correlations.csv', index_col=0)
print(f"\n🔗 ASSET CORRELATION INSIGHTS:")
print("   Assets most positively correlated with portfolio performance:")
top_positive_corr = asset_correlations_df.nlargest(3, 'Sharpe_Corr')
for _, row in top_positive_corr.iterrows():
    print(f"   • {row['Asset']}: Sharpe correlation {row['Sharpe_Corr']:.3f}, Avg allocation {row['Avg_Allocation']:.1%}")

print("\n   Assets most negatively correlated with portfolio performance:")
top_negative_corr = asset_correlations_df.nsmallest(3, 'Sharpe_Corr')
for _, row in top_negative_corr.iterrows():
    print(f"   • {row['Asset']}: Sharpe correlation {row['Sharpe_Corr']:.3f}, Avg allocation {row['Avg_Allocation']:.1%}")

fascinating_insights['diversification_insights'] = {
    'positive_performers': top_positive_corr[['Asset', 'Sharpe_Corr', 'Avg_Allocation']].to_dict('records'),
    'negative_performers': top_negative_corr[['Asset', 'Sharpe_Corr', 'Avg_Allocation']].to_dict('records')
}

# Cost efficiency analysis
print(f"\n💰 COST EFFICIENCY ANALYSIS:")
cost_range = pareto_results_df['cost'].max() - pareto_results_df['cost'].min()
avg_cost = pareto_results_df['cost'].mean()
traditional_avg_cost = bogleheads_comparison_df['cost'].mean()

print(f"   Pareto portfolio costs: {pareto_results_df['cost'].min():.3%} - {pareto_results_df['cost'].max():.3%}")
print(f"   Average Pareto cost: {avg_cost:.3%}")
print(f"   Average Traditional cost: {traditional_avg_cost:.3%}")
print(f"   Cost difference: {(traditional_avg_cost - avg_cost)*100:.2f} basis points")

fascinating_insights['cost_efficiency'] = {
    'pareto_cost_range': [pareto_results_df['cost'].min(), pareto_results_df['cost'].max()],
    'pareto_avg_cost': avg_cost,
    'traditional_avg_cost': traditional_avg_cost,
    'cost_advantage_bps': (traditional_avg_cost - avg_cost) * 10000
}

# International exposure analysis
print(f"\n🌍 INTERNATIONAL EXPOSURE ANALYSIS:")
international_assets = ['VXUS', 'VTIAX', 'BNDX', 'VSS']
total_international = pareto_portfolios_df[international_assets].sum(axis=1)
avg_international = total_international.mean()
max_international = total_international.max()
portfolios_with_international = (total_international > 0.01).sum()

print(f"   Average international allocation: {avg_international:.1%}")
print(f"   Maximum international allocation: {max_international:.1%}")
print(f"   Portfolios with meaningful int'l exposure (>1%): {portfolios_with_international}/{len(pareto_portfolios_df)} ({portfolios_with_international/len(pareto_portfolios_df):.1%})")

# US equity concentration analysis
us_equity_assets = ['VOO', 'VTI', 'VTV', 'VBR']
total_us_equity = pareto_portfolios_df[us_equity_assets].sum(axis=1)
avg_us_equity = total_us_equity.mean()
print(f"   Average US equity allocation: {avg_us_equity:.1%}")

fascinating_insights['geographic_allocation'] = {
    'avg_international': avg_international,
    'max_international': max_international,
    'international_usage_rate': portfolios_with_international/len(pareto_portfolios_df),
    'avg_us_equity': avg_us_equity
}

print(f"\n✅ Comprehensive analysis complete - extracted {len(fascinating_insights)} categories of insights!")
print(f"🎯 Ready to generate enhanced reports with these fascinating findings!")
````
